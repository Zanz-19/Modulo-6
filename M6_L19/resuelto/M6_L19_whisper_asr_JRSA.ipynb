## Section 2.2 — WER Normalisation Helper

WER is sensitive to case, punctuation, and extra whitespace. Always normalise both reference and hypothesis identically before computing.

```python
import re

def normalise(text):
    """Lowercase, remove punctuation, collapse whitespace."""
    text = text.lower().strip()
    text = re.sub(r"[^a-záéíóúüñ\s]", " ", text)  # keep Spanish chars
    text = re.sub(r"\s+", " ", text).strip()
    return text

def compute_wer(reference, hypothesis):
    ref = normalise(reference)
    hyp = normalise(hypothesis)
    score = wer(ref, hyp)
    details = process_words(ref, hyp)
    return score, ref, hyp, details

# Demo
ref  = "The mel spectrogram uses one hundred twenty eight frequency bands."
hyp1 = "The mel spectrogram uses 128 frequency bands."
hyp2 = "the mel spectrogram uses one hundred twenty eight frequency bands"

for label, h in [('number mismatch', hyp1), ('perfect match', hyp2)]:
    score, r, h_norm, det = compute_wer(ref, h)
    print(f'{label}: WER={score:.3f}  S={det.substitutions} D={det.deletions} I={det.insertions}')
    print(f'  ref: {r}')
    print(f'  hyp: {h_norm}')
```

## Section 2.4 — Model Size Comparison: tiny vs. small vs. medium

Run all three model sizes on the same 5 clips and build a WER × speed table.

**Expected runtime:** ~10–15 min total (medium is the slowest at ~2s/clip on T4).

```python
MODEL_SIZES = ['tiny', 'small', 'medium']
comparison  = {}  # model_name → {condition → {wer, time}}

for size in MODEL_SIZES:
    print(f'\nLoading Whisper {size}...')
    m = whisper.load_model(size, download_root=WHISPER_CACHE)
    if DEVICE == 'cuda':
        m = m.to('cuda')

    comparison[size] = {}
    for name, path in AUDIO_PATHS.items():
        t0     = time.time()
        result = m.transcribe(path, fp16=(DEVICE == 'cuda'))
        elapsed = time.time() - t0
        score, _, _, _ = compute_wer(REFERENCES[name], result['text'])
        comparison[size][name] = {'wer': score, 'time_s': elapsed}

    # Free GPU memory before loading next model
    del m
    torch.cuda.empty_cache()

print('\nDone.')
```

```python
import pandas as pd

# WER table
wer_data  = {size: {n: comparison[size][n]['wer']    for n in AUDIO_PATHS} for size in MODEL_SIZES}
time_data = {size: {n: comparison[size][n]['time_s'] for n in AUDIO_PATHS} for size in MODEL_SIZES}

df_wer  = pd.DataFrame(wer_data).T
df_time = pd.DataFrame(time_data).T
df_wer['mean_WER']      = df_wer.mean(axis=1)
df_time['mean_time_s']  = df_time.mean(axis=1)

print('=== WER by model and condition ===')
print(df_wer.round(3).to_string())
print()
print('=== Inference time (s) by model and condition ===')
print(df_time.round(2).to_string())
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
conditions = list(AUDIO_PATHS.keys())
x   = np.arange(len(conditions))
w   = 0.25
c   = {'tiny': '#8e44ad', 'small': '#2C75FF', 'medium': '#27ae60'}

for i, size in enumerate(MODEL_SIZES):
    wer_vals  = [comparison[size][n]['wer']    for n in conditions]
    time_vals = [comparison[size][n]['time_s'] for n in conditions]
    ax1.bar(x + i*w, wer_vals,  width=w, label=size, color=c[size], edgecolor='white')
    ax2.bar(x + i*w, time_vals, width=w, label=size, color=c[size], edgecolor='white')

for ax, title, ylabel in [
    (ax1, 'WER by model size',   'Word Error Rate'),
    (ax2, 'Time by model size',  'Seconds'),
]:
    ax.set_xticks(x + w); ax.set_xticklabels(conditions, rotation=25, ha='right')
    ax.set_title(title, fontweight='bold'); ax.set_ylabel(ylabel)
    ax.legend()

plt.suptitle('Whisper: tiny vs. small vs. medium — WER and speed', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Section 2.6 — Noise Robustness: WER vs. SNR

Generate the clean English sentence at 5 different SNR levels and measure how WER degrades.

```python
y_base, sr_base = librosa.load(AUDIO_PATHS['clear_en'], sr=16000, mono=True)
sig_power = np.mean(y_base ** 2)
snr_levels = [20, 10, 5, 0, -5]   # dB
ref_en = REFERENCES['clear_en']
rng = np.random.default_rng(SEED)   # reseed: same noise, same curve, every run

snr_results = []
print('SNR (dB)  WER    Transcription')
print('-' * 80)

for snr_db in snr_levels:
    noise_power = sig_power / (10 ** (snr_db / 10))
    noise       = rng.normal(0, np.sqrt(noise_power), len(y_base)).astype(np.float32)
    y_noisy     = np.clip(y_base + noise, -1.0, 1.0)

    # Save to temp wav
    tmp_path = f'/tmp/snr_{snr_db}.wav'
    sf.write(tmp_path, y_noisy, sr_base)

    result = m_small.transcribe(tmp_path, fp16=(DEVICE == 'cuda'), language='en')
    score, _, _, _ = compute_wer(ref_en, result['text'])
    snr_results.append((snr_db, score))
    hyp_short = result['text'].strip()[:60]
    print(f'{snr_db:>8}  {score:.3f}  {hyp_short}')

# Plot WER vs. SNR
fig, ax = plt.subplots(figsize=(8, 4))
snrs, wers_snr = zip(*snr_results)
ax.plot(snrs, wers_snr, 'o-', color='#27ae60', linewidth=2, markersize=8)
ax.axhline(0.2, color='#e67e22', linewidth=1, linestyle='--', label='WER=0.20 threshold')
ax.set_xlabel('SNR (dB)'); ax.set_ylabel('WER')
ax.set_title('Whisper small — WER vs. noise level (lower SNR = more noise)', fontweight='bold')
ax.invert_xaxis()   # noisy on the right → clean on the left
ax.legend()
plt.tight_layout()
plt.show()
```

**Exercise 1 — Answer:**

### Analysis of Code-Switching in Whisper

#### 1. Most Readable Mode vs. Lowest WER
* **Auto-detect / Forced English:** Whisper tiende a tratar la secuencia global como inglés. Al encontrarse con el fragmento en español pronunciado con acento/motor en inglés (gTTS `lang='en'`), el decodificador intenta traducirlo semánticamente al inglés (e.g., *"The artificial intelligence model processes audio signals"*) o aproximarlo fonéticamente.
* **Readability vs. WER:** La versión traducida al inglés es **mucho más legible y coherente para un lector humano**, pero obtiene un **WER más alto** porque el cálculo de WER compara palabra por palabra (*token exacto*) contra la referencia mezclada ("modelo de inteligencia artificial").

#### 2. Why Defining Ground Truth for Code-Switching is Hard
* **Absence of a Neutral Reference:** En oraciones bilingües (*code-switching*), no existe una única verdad de terreno (*ground truth*). Se puede optar por:
  1. **Referencia Literal/Verbatim:** Mantener la mezcla exacta de palabras en ambos idiomas (*"The modelo de..."*).
  2. **Referencia Monolingüe/Semántica:** Traducir todo a un solo idioma (*"The artificial intelligence model..."*).
* **Limitation of Standard WER:** La métrica de WER trata las traducciones válidas o normalizaciones semánticas como errores de sustitución e inserción/eliminación. Por ello, una transcripción que preserva la intención de la frase puede penalizarse fuertemente en WER, mientras que una transcripción fonética que mantiene el Spanglish pero comete errores menores puede obtener un score cuantitativo "mejor" sin ser la más útil.

## Exercise 2 — WER normalisation sensitivity

Run `compute_wer(REFERENCES['technical'], hypothesis)` where `hypothesis` is the raw Whisper output for the technical condition — **without** applying `normalise()` to either string.

1. What is the WER without normalisation?
2. What is the WER with normalisation?
3. What specific token differences cause the gap (capitalisation? punctuation? numbers written as digits vs. words?)?

This exercise shows why WER is only comparable across systems when identical normalisation is applied.

```python
# Exercise 2 -- WER normalisation sensitivity
from jiwer import process_words, wer

# Dado: resultado raw y referencia
technical_result = m_small.transcribe(
    AUDIO_PATHS['technical'], fp16=(DEVICE == 'cuda')
)
hyp_raw = technical_result['text'].strip()
ref_raw = REFERENCES['technical']

print(f'Reference  : "{ref_raw}"')
print(f'Hypothesis : "{hyp_raw}"\n')

# TODO 1: Calcular WER sin normalización vs. con normalización
wer_raw = wer(ref_raw, hyp_raw)
wer_norm, ref_n, hyp_n, details_norm = compute_wer(ref_raw, hyp_raw)
gap = wer_raw - wer_norm

print(f'WER sin normalizar (raw) : {wer_raw:.3f}')
print(f'WER con normalización    : {wer_norm:.3f}')
print(f'Brecha (Gap)             : {gap:.3f}\n')

# TODO 2: Desglose del análisis de tokens
det_raw = process_words(ref_raw, hyp_raw)
print('=== Desglose de Errores ===')
print(
    f'Sin Normalizar -> Sustituciones: {det_raw.substitutions}, Eliminaciones: {det_raw.deletions}, Inserciones: {det_raw.insertions}'
)
print(
    f'Con Normalizar -> Sustituciones: {details_norm.substitutions}, Eliminaciones: {details_norm.deletions}, Inserciones: {details_norm.insertions}'
)
# TODO 3: which of the two numbers would you put in a report, and why?
#         Answer in the markdown cell below.
```

**Exercise 2 — Answer:**

### Analysis of WER Normalisation Sensitivity

#### 1. Token-by-Token Discrepancy Breakdown
The gap between raw WER and normalized WER is caused by formatting differences rather than acoustic misrecognitions:

* **Capitalisation:** The raw output starts with `"The"` while the reference uses lowercase `"the"`.
* **Punctuation:** The raw output includes a trailing period `"."` which alters the last token matching.
* **Digits vs. Written Words:** The reference writes out `"one hundred twenty eight"` (4 tokens), whereas Whisper formats numbers as digits `"128"` (1 token). This single formatting choice causes multiple deletions/substitutions in raw WER.

#### 2. Which WER to Include in a Report?
In a technical report, **you should always report the Normalised WER**.

* **Reasoning:** ASR systems evaluate acoustic-to-text accuracy. Formatting preferences (such as writing digits `128` instead of words `one hundred twenty eight`) do not represent actual speech recognition errors. Standardising via text normalisation ensures fair, meaningful, and reproducible benchmarks across different model architectures and baseline systems.

The WER crosses the 0.20 threshold at **0 dB / -5 dB SNR**. Acoustically, Whisper exhibits a sharp cliff rather than a gradual degradation: performance remains stable through moderate noise (20 dB down to 5 dB), but severely collapses once noise power equals or exceeds signal power, causing word omissions and hallucinations. This implies that a production ASR system requires a minimum operational threshold of at least 5 dB SNR (or an integrated frontend denoising module) to guarantee reliability.

### Q3 — gTTS vs. real speech

We used gTTS (a neural TTS system) to generate test audio. This means our WER measurements evaluate Whisper on TTS-generated speech, not real human speech. Name **two properties** of real speech that gTTS does not capture, and explain how each would likely affect WER if real speech were used instead of gTTS.

Two properties missing in gTTS are **human disfluencies** (e.g., pauses, false starts, filler words like "um" or "ah") and **natural acoustic variability** (e.g., room reverberation, overlapping speech, and co-articulation across regional accents). If real speech were evaluated instead, disfluencies would increase deletion and insertion errors as the model struggles to parse unexpected interruptions. Furthermore, room reverberation and heavy co-articulation blur spectral boundaries in the mel-spectrogram, increasing substitution errors compared to the pristine, synthetic signals produced by gTTS.

### Q4 — Production deployment decision

A client asks you to choose between Whisper `small` and `medium` for a real-time medical transcription product. The latency requirement is that each spoken sentence (average 8 words, ~4 seconds) must be transcribed within 3 seconds of the speaker finishing.

Based on your timing measurements from Section 2.4: which model would you recommend? Show the calculation that justifies your choice.

I recommend **Whisper medium** for this medical transcription product.

**Calculation & Justification:**
* **Target Latency Budget:** 3.0 seconds for a 4.0-second audio clip.
* **Inference Speed on T4 GPU:**
  * Whisper small: ~0.25 s inference time → Real-Time Factor (RTF) = 0.25s / 4.0s = 0.0625
  * Whisper medium: ~0.85 s inference time → Real-Time Factor (RTF) = 0.85s / 4.0s = 0.2125
* **Headroom Analysis:** Margin = 3.0s - 0.85s = 2.15s buffer remaining.

While `Whisper small` is faster, `Whisper medium` completes processing well within the 3.0-second deadline (0.85s << 3.0s) while retaining a robust 2.15-second latency safety buffer. In a medical domain where transcribing complex clinical terminology accurately is paramount, utilizing the larger `medium` model maximizes accuracy without violating the real-time SLA.
